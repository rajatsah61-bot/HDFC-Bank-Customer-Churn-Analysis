```python
country_count = df["country"].value_counts()
country_count
```

```python
plt.figure(figsize=(8,5))
plt.bar(country_count.index, country_count.values)

plt.title("Customers by Country")
plt.xlabel("Country")
plt.ylabel("Number of Customers")

plt.savefig("../Images/customers_by_country.png")

plt.show()
```

```python
gender = df["gender"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(gender.index, gender.values)
plt.title("Customers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.savefig("../Images/customers_by_Gender.png")
plt.show()
```

```python
churn = df["churn"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(churn.index, churn.values)
plt.title("Churn Distribution")
plt.xlabel("Churn (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Customers")
plt.savefig("../Images/customers_by_Number of Customers.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["credit_score"], bins=20)
plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Number of Customers")
plt.savefig("../Images/customers_by_Number of Customers.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.savefig("../Images/customers_by_Age.png")
plt.show()
```

```python
country_churn = df.groupby("country")["churn"].sum()
print(country_churn)
```

```python
plt.figure(figsize=(8,5))
plt.bar(country_churn.index, country_churn.values)
plt.title("Customer Churn by Country")
plt.xlabel("Country")
plt.ylabel("Customers Left")
plt.show()
```

```python
gender_churn = df.groupby("gender")["churn"].sum()
print(gender_churn)
```

```python
plt.figure(figsize=(6,4))
plt.bar(gender_churn.index, gender_churn.values)
plt.title("Churn by Gender")
plt.xlabel("Gender")
plt.ylabel("Customers Left")
plt.show()
```

```python
df["Age_Group"] = pd.cut(
    df["age"],
    bins=[18,30,40,50,60,100],
    labels=["18-30","31-40","41-50","51-60","60+"]
)
```

```python
age_churn = df.groupby("Age_Group")["churn"].sum()
print(age_churn)
```

```python
plt.figure(figsize=(8,5))
plt.bar(age_churn.index.astype(str), age_churn.values)
plt.title("Churn by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Customers Left")
plt.savefig("../Images/customers_by_Age Group.png")
plt.show()
```

```python
df.groupby("churn")["balance"].mean()
```

```python
salary = df.groupby("country")["estimated_salary"].mean()
print(salary)
```

```python
plt.figure(figsize=(8,5))
plt.bar(salary.index, salary.values)
plt.title("Average Salary by Country")
plt.xlabel("Country")
plt.ylabel("Average Salary")
plt.show()
```

```python
credit = df.groupby("country")["credit_score"].mean()
print(credit)
```

```python
plt.figure(figsize=(8,5))
plt.bar(credit.index, credit.values)

plt.title("Average Credit Score by Country")
plt.xlabel("Country")
plt.ylabel("Credit Score")

plt.show()
```

```python
active = df["active_member"].value_counts()
print(active)
```

```python
plt.figure(figsize=(6,4))
plt.bar(active.index.astype(str), active.values)

plt.title("Active vs Inactive Members")
plt.xlabel("Active Member (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")
plt.savefig("../Images/customers_by_Active Member (0 = No, 1 = Yes).png")
plt.show()
```

```python
country = df["country"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(country.values,
        labels=country.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Customer Distribution by Country")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["balance"])
plt.title("Customer Balance")
plt.ylabel("Balance")
plt.savefig("../Images/customer_balance_boxplot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["age"], df["balance"])
plt.title("Age vs Balance")
plt.xlabel("Age")
plt.ylabel("Balance")
plt.savefig("../Images/customers_by_Age.png")
plt.show()
```
